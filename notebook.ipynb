```python
import pandas as pd
import sqlite3

# Загрузка исходных данных
df = pd.read_csv(r'D:\ecommerce-project\data\data.csv\data.csv', encoding='latin1')

print(f"Исходное количество строк: {len(df)}")

# Правило 1 — удаляем явные служебные StockCode
service_codes = ['D', 'POST', 'M', 'BANK CHARGES', 'C2', 'DOT', 'CRUK', 'PADS', 'S']
df_clean = df[~df['StockCode'].isin(service_codes)].copy()

# Правило 2 — удаляем служебные записи, спрятанные в Description
service_descriptions = ['check', '?', 'damages', 'damaged', 'Damaged', 'found', 'Found',
                          'adjustment', 'thrown away', 'Unsaleable, destroyed.',
                          'amazon', 'sold as set on dotcom']
df_clean = df_clean[~df_clean['Description'].isin(service_descriptions)]

# Правило 3 — удаляем "мусорные" строки: нулевая цена И нет описания
df_clean = df_clean[~((df_clean['UnitPrice'] == 0) & (df_clean['Description'].isnull()))]

# Правило 4 — убираем оставшиеся строки с нулевой или отрицательной ценой
# (возвраты с отрицательным Quantity остаются — это легитимные операции)
df_clean = df_clean[df_clean['UnitPrice'] > 0]

# Правило 5 — конвертация типов
df_clean['InvoiceDate'] = pd.to_datetime(df_clean['InvoiceDate'])

# Новый столбец — явный признак возврата (для будущего анализа)
df_clean['IsReturn'] = df_clean['InvoiceNo'].astype(str).str.startswith('C')

# Новый столбец — выручка по строке (понадобится почти для всех вопросов)
df_clean['Revenue'] = df_clean['Quantity'] * df_clean['UnitPrice']

print(f"После очистки: {len(df_clean)}")
print(f"Удалено строк: {len(df) - len(df_clean)}")

df_clean.head()
```

```python
conn = sqlite3.connect(r'D:\ecommerce-project\ecommerce.db')
df_clean.to_sql('transactions', conn, if_exists='replace', index=False)
conn.close()

print("Данные записаны в SQLite")
```

```python
import pandas as pd
import sqlite3

conn = sqlite3.connect(r'D:\ecommerce-project\ecommerce.db')
df_clean = pd.read_sql("SELECT * FROM transactions", conn, parse_dates=['InvoiceDate'])
conn.close()

print(f"Загружено строк: {len(df_clean)}")
print(df_clean.dtypes)
```

```python
# Только клиенты с ID и только покупки (без возвратов)
df_rfm_base = df_clean[df_clean['CustomerID'].notnull()].copy()
purchases = df_rfm_base[df_rfm_base['IsReturn'] == 0]

snapshot_date = purchases['InvoiceDate'].max() + pd.Timedelta(days=1)

rfm = purchases.groupby('CustomerID').agg(
    Recency=('InvoiceDate', lambda x: (snapshot_date - x.max()).days),
    Frequency=('InvoiceNo', 'nunique'),
    Monetary=('Revenue', 'sum')
).reset_index()

print(f"Клиентов в RFM: {len(rfm)}")
print(f"Точка отсчёта: {snapshot_date}")
print(rfm.head())
print(rfm.describe())
```

```python
# Чистая выручка клиента: покупки + возвраты (возвраты отрицательные)
net_monetary = (df_rfm_base.groupby('CustomerID')['Revenue']
                .sum().rename('Monetary').reset_index())

rfm = rfm.drop(columns='Monetary').merge(net_monetary, on='CustomerID')

# Клиенты, у которых чистая выручка <= 0 (всё вернули), в RFM не нужны
before = len(rfm)
rfm = rfm[rfm['Monetary'] > 0.01].copy()
print(f"Исключено клиентов с нулевой/отрицательной чистой выручкой: {before - len(rfm)}")
print(rfm['Monetary'].describe())
```

```python
# Recency: чем МЕНЬШЕ дней, тем лучше, поэтому метки в обратном порядке
rfm['R'] = pd.qcut(rfm['Recency'], 5, labels=[5, 4, 3, 2, 1]).astype(int)

# Frequency: много одинаковых значений (1, 2...), поэтому сначала rank(method='first'),
# иначе qcut упадёт с ошибкой про повторяющиеся границы
rfm['F'] = pd.qcut(rfm['Frequency'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5]).astype(int)

rfm['M'] = pd.qcut(rfm['Monetary'], 5, labels=[1, 2, 3, 4, 5]).astype(int)

def segment(row):
    if row['R'] >= 4 and row['F'] >= 4:
        return 'Чемпионы'
    elif row['F'] >= 4:
        return 'Лояльные, остывающие'
    elif row['R'] >= 4:
        return 'Недавние'
    elif row['R'] <= 2 and row['F'] <= 2:
        return 'Потерянные'
    else:
        return 'Середина'

rfm['Segment'] = rfm.apply(segment, axis=1)

summary = rfm.groupby('Segment').agg(
    clients=('CustomerID', 'count'),
    avg_recency=('Recency', 'mean'),
    avg_frequency=('Frequency', 'mean'),
    total_revenue=('Monetary', 'sum')
).round(1)
summary['pct_revenue'] = (summary['total_revenue'] / summary['total_revenue'].sum() * 100).round(1)
print(summary.sort_values('total_revenue', ascending=False))
```

```python
# R: чем меньше дней с последней покупки, тем выше балл
rfm['R'] = pd.qcut(rfm['Recency'], 5, labels=[5, 4, 3, 2, 1]).astype(int)

# F: явные пороги, чтобы результат не зависел от порядка строк
def f_score(n):
    if n == 1:   return 1
    elif n == 2: return 2
    elif n <= 4: return 3
    elif n <= 7: return 4
    else:        return 5

rfm['F'] = rfm['Frequency'].apply(f_score)

# M: квинтили по чистой выручке
rfm['M'] = pd.qcut(rfm['Monetary'], 5, labels=[1, 2, 3, 4, 5]).astype(int)

def segment(row):
    if row['R'] >= 4 and row['F'] >= 4:
        return 'Чемпионы'
    elif row['F'] >= 4:
        return 'Лояльные, остывающие'
    elif row['R'] >= 4:
        return 'Недавние'
    elif row['R'] <= 2 and row['F'] <= 2:
        return 'Потерянные'
    else:
        return 'Середина'

rfm['Segment'] = rfm.apply(segment, axis=1)

summary = rfm.groupby('Segment').agg(
    clients=('CustomerID', 'count'),
    avg_recency=('Recency', 'mean'),
    avg_frequency=('Frequency', 'mean'),
    total_revenue=('Monetary', 'sum')
).round(1)
summary['pct_revenue'] = (summary['total_revenue'] / summary['total_revenue'].sum() * 100).round(1)
print(summary.sort_values('total_revenue', ascending=False))
```

```python
import sqlite3

conn = sqlite3.connect(r'D:\ecommerce-project\ecommerce.db')
rfm.to_sql('rfm', conn, if_exists='replace', index=False)
conn.close()
rfm.to_csv(r'D:\ecommerce-project\data\rfm_segments.csv', index=False)
print("rfm сохранён:", len(rfm), "клиентов")
```
